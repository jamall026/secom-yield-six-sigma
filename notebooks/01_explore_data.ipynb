```python
import pandas as pd
from ucimlrepo import fetch_ucirepo
```

```python
secom = fetch_ucirepo(id=179)
df = secom.data.original
```

```python
print("Full table:", df.shape)
print("First 5 columns:", df.columns.tolist()[:5])
print("Last 5 columns:", df.columns.tolist()[-5:])
```

```python
y = df["class"]
timestamps = df["timestamp"]
X = df.drop(columns=["class", "timestamp"])

print("Sensor readings (X):", X.shape)
print("Results (y):", y.shape)
```

```python
X.head()
```

```python
counts = y.value_counts()
print(counts)

passed = (y == -1).sum()
failed = (y == 1).sum()
yield_pct = passed / len(y) * 100

print(f"\nPassed: {passed}, Failed: {failed}")
print(f"Yield: {yield_pct:.1f}%")
```

```python
dates = pd.to_datetime(timestamps, format="mixed", dayfirst=True)

print("First unit tested:", dates.min())
print("Last unit tested: ", dates.max())
print("Days covered:", (dates.max() - dates.min()).days)
```

```python
days = 89
annual_factor = 365 / days

annual_units = len(y) * annual_factor
annual_fails = failed * annual_factor

print(f"Units per year (estimate): {annual_units:,.0f}")
print(f"Failed units per year (estimate): {annual_fails:,.0f}")
print()

for cost in [1000, 3000, 5000]:
    loss = annual_fails * cost
    saving = loss / 2
    print(f"At ${cost:,} per failed unit: loss ${loss:,.0f}/year, "
          f"halving failures saves ${saving:,.0f}/year")
```
